```python
# 랜덤하게 1000명의 키와 몸무게를 만들어서 파일로 저장하기

import random
random.seed(1)
last_names = list("김이박최정강조윤장임오한신서권황안송류전홍고문양손배백허유남심노하주하")
first_names = list("가강건경고관광규근기길나남노누다단달담대덕도동두라래로루리마만명무미민바백범별보빛사산상서석선설성세소솔수순승시신아안애양어연영예오옥완요우윤율은의이익인일자재전정제조주준지진찬창채천철초춘충태택하한해현호홍화환회휘희")

def findBmi(weight, height):
    bmi = weight / ((height / 100) ** 2)
    if bmi >= 30:
        return bmi, "비만"
    elif bmi >= 25:
        return bmi, "과체중"
    elif bmi >= 18.5:
        return bmi,  "정상"
    else:
        return bmi, "저체중"

with open("info.csv", "w", encoding='utf-8') as f:
    for i in range(10):
        name = random.choice(last_names) + ''.join(random.choices(first_names, k=2))
        weight = random.randrange(40, 100)
        height = random.randrange(140, 200)
        f.write(f"{name}, {weight},{height}\n")


with open("info.csv", "r", encoding='utf-8') as f:
    for line in f:
        name, weight, height = line.strip().split(",")
        bmi, bmi_category = findBmi(int(weight), int(height))
        print(f"이름: {name}, 몸무게: {weight}kg, 키: {height}cm, BMI: {bmi:.2f}, 비만도: {bmi_category}")
```

```python
from urllib.request import urlopen
url = "https://www.gutenberg.org/files/11/11-0.txt"

with open("alice.txt", "w", encoding='utf-8') as f:
  with urlopen(url) as story:
      for line in story:
          f.write(line.decode('utf-8'))

with open("alice.txt", "r", encoding='utf-8') as f:
    story_words = []
    for line in f:
        line_words = line.split()
        
        for word in line_words:
            story_words.append(word)
    print(story_words[:100])
```
